# 02. Exploratory Data Analysis (EDA)
### Exploring Text Lengths, Class Balance, and Word Frequencies in Indic Climate Discourse


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

import pandas as pd
import matplotlib.pyplot as plt
from src.data.dataset_loader import load_custom

records = load_custom()
df = pd.DataFrame(records)
df['char_length'] = df['text'].apply(len)
df['word_count'] = df['text'].apply(lambda x: len(x.split()))
```

## 1. Sequence Length Statistics


```python
display(df.groupby('language')[['char_length', 'word_count']].describe().round(1))
```

## 2. Distribution Across Scripts


```python
display(df.groupby(['language', 'script', 'label']).size().unstack(fill_value=0))
```
